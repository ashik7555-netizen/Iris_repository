```python
def get_product_price(product_id, products_list):
    """
    Returns the Price of the matching product from the products_list,
    or None if the product ID is not found.
    """
    for product in products_list:
        if product['Product_ID'] == product_id:
            return product['Price']
    return None

# Example usage:
price_p01 = get_product_price('P01', products)
print(f"Price of P01: ${price_p01}")

price_p99 = get_product_price('P99', products)
print(f"Price of P99 (non-existent): {price_p99}")
```

```python
def get_customer_name(customer_id, customers_list):
    """
    Returns the Name of the matching customer from the customers_list,
    or None if the Customer_ID is not found.
    """
    for customer in customers_list:
        if customer['Customer_ID'] == customer_id:
            return customer['Name']
    return None

# Example usage:
name_c01 = get_customer_name('C01', customers)
print(f"Name for C01: {name_c01}")

name_c99 = get_customer_name('C99', customers)
print(f"Name for C99 (non-existent): {name_c99}")
```

```python
def calculate_order_value(price, quantity):
    """
    Calculates the total value of an order based on price and quantity.
    """
    return price * quantity

# Example usage:
price_per_item = 250
items_ordered = 3
total_value = calculate_order_value(price_per_item, items_ordered)
print(f"The total order value for {items_ordered} items at ${price_per_item} each is: ${total_value}")
```

```python
def classify_order(value, high_limit=3000, low_limit=1000):
    """
    Classifies an order as 'High', 'Medium', or 'Low' based on its value.
    """
    if value >= high_limit:
        return "High"
    elif value >= low_limit:
        return "Medium"
    else:
        return "Low"

# Example usage with default thresholds:
order_value1 = 3500
order_value2 = 1500
order_value3 = 500

print(f"Order value ${order_value1}: {classify_order(order_value1)}")
print(f"Order value ${order_value2}: {classify_order(order_value2)}")
print(f"Order value ${order_value3}: {classify_order(order_value3)}")

# Example usage with custom thresholds:
print(f"Order value ${order_value1} (custom limits): {classify_order(order_value1, high_limit=4000, low_limit=2000)}")
```

```python
def generate_invoice(order, products_list, customers_list):
    """
    Generates a neatly formatted invoice string for a single order.
    Assumes `get_product_price`, `get_customer_name`, `calculate_order_value`,
    and `classify_order` functions are defined in the environment.
    """

    order_id = order['Order_ID']
    customer_id = order['Customer_ID']
    product_id = order['Product_ID']
    quantity = order['Quantity']
    payment_mode = order['Payment_Mode']
    status = order['Status']

    # Use previously defined functions
    customer_name = get_customer_name(customer_id, customers_list)
    product_price = get_product_price(product_id, products_list)

    # Ensure product_price is not None before calculating value
    if product_price is not None:
        order_value = calculate_order_value(product_price, quantity)
        order_classification = classify_order(order_value)
    else:
        order_value = 0
        order_classification = "N/A"

    # Find product name (assuming product_id is valid for display)
    product_name = "Unknown Product"
    for product in products_list:
        if product['Product_ID'] == product_id:
            product_name = product['Name']
            break

    invoice_string = """
--- Invoice Details for Order #{order_id} ---
Customer Name: {customer_name}
Product: {product_name} (ID: {product_id})
Quantity: {quantity}
Unit Price: ${product_price:.2f}
--------------------------------------
Total Order Value: ${order_value:.2f}
Order Classification: {order_classification}
Payment Mode: {payment_mode}
Order Status: {status}
---------------------------------------
""".format(
        order_id=order_id,
        customer_name=customer_name,
        product_name=product_name,
        product_id=product_id,
        quantity=quantity,
        product_price=product_price if product_price is not None else 0,
        order_value=order_value,
        order_classification=order_classification,
        payment_mode=payment_mode,
        status=status
    )
    return invoice_string

# Example usage with an order from the 'orders' list:
example_order = orders[0] # Taking the first order as an example
invoice = generate_invoice(example_order, products, customers)
print(invoice)

example_order_cancelled = orders[4] # Order 0105 which is cancelled
invoice_cancelled = generate_invoice(example_order_cancelled, products, customers)
print(invoice_cancelled)
```

```python
def calculate_total_revenue(orders_list, products_list):
    """
    Calculates the total revenue from a list of orders, skipping cancelled ones.
    Assumes `get_product_price` function is defined in the environment.
    """
    total_revenue = 0
    for order in orders_list:
        # Skip cancelled orders
        if order['Status'] == 'Cancelled':
            continue

        # Use the previously defined get_product_price function
        product_price = get_product_price(order['Product_ID'], products_list)

        if product_price is not None:
            order_value = product_price * order['Quantity']
            total_revenue += order_value
    return total_revenue

# Call the function and confirm it matches the Module 4 answer
calculated_revenue = calculate_total_revenue(orders, products)
module4_revenue = 28184 # The answer from Module 4

print(f"Calculated Total Revenue: ${calculated_revenue}")
print(f"Module 4 Total Revenue: ${module4_revenue}")

if calculated_revenue == module4_revenue:
    print("Confirmation: The calculated revenue matches the Module 4 answer.")
else:
    print("Warning: The calculated revenue does NOT match the Module 4 answer.")
```

```python
def top_customer_by_spend(orders_list, products_list, customers_list):
    """
    Returns the Name of the customer who has spent the most money (Delivered orders only).
    Assumes `get_product_price` and `get_customer_name` functions are defined.
    """
    customer_spend = {}

    for order in orders_list:
        if order['Status'] == 'Delivered':
            customer_id = order['Customer_ID']
            product_id = order['Product_ID']
            quantity = order['Quantity']

            product_price = get_product_price(product_id, products_list)

            if product_price is not None:
                order_value = product_price * quantity
                customer_spend[customer_id] = customer_spend.get(customer_id, 0) + order_value

    if not customer_spend:
        return None # No delivered orders or no spending found

    # Find the customer with the maximum spend
    top_customer_id = max(customer_spend, key=customer_spend.get)

    # Get the name of the top customer using the existing function
    top_customer_name = get_customer_name(top_customer_id, customers_list)

    return top_customer_name

# Example usage:
top_customer = top_customer_by_spend(orders, products, customers)
if top_customer:
    print(f"The customer with the highest spend (delivered orders only) is: {top_customer}")
else:
    print("No top customer found.")
```

```python
def apply_offer(price, *discounts):
    """
    Applies multiple discount percentages sequentially to a price.

    Args:
        price (float or int): The initial price.
        *discounts (float or int): Variable number of discount percentages (e.g., 10 for 10%).

    Returns:
        float: The final discounted price.
    """
    current_price = float(price)
    for discount in discounts:
        # Ensure discount is a percentage (e.g., 10 for 10%, not 0.10)
        current_price *= (1 - discount / 100)
    return current_price

# Example usage:
initial_price = 1000
discount1 = 10 # 10% discount
discount2 = 5  # 5% discount
discount3 = 20 # 20% discount

final_price_single_discount = apply_offer(initial_price, discount1)
print(f"Initial Price: ${initial_price:.2f}")
print(f"After {discount1}% discount: ${final_price_single_discount:.2f}")

final_price_multiple_discounts = apply_offer(initial_price, discount1, discount2)
print(f"After {discount1}% and then {discount2}% discounts: ${final_price_multiple_discounts:.2f}")

final_price_all_discounts = apply_offer(initial_price, discount1, discount2, discount3)
print(f"After {discount1}%, {discount2}%, and then {discount3}% discounts: ${final_price_all_discounts:.2f}")

no_discount_price = apply_offer(initial_price)
print(f"With no discounts: ${no_discount_price:.2f}")
```
